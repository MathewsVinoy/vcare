```python
data_path1 = "data/blood/biased_leukemia_dataset.csv"
data_path2 = "data/blood/The_Cancer_data_1500_V2.csv"
```

```python
import pandas as pd

df1 = pd.read_csv("data/blood/biased_leukemia_dataset.csv")
df2 = pd.read_csv(data_path2)
df1.head()
```

```python
print("Gender:", df1['Gender'].unique())
print("Smoking_Status:", df1['Smoking_Status'].unique())
print("Genetic_Mutation:", df1['Genetic_Mutation'].unique())
print("Family_History:", df1['Family_History'].unique())
print("Radiation_Exposure:", df1['Radiation_Exposure'].unique())
print("Infection_History:", df1['Infection_History'].unique())
print("Leukemia_Status:", df1['Leukemia_Status'].unique())

df1['Gender'] = df1['Gender'].map({'Male': 1, 'Female': 0}).fillna(df1['Gender'])
df1['Smoking_Status'] = df1['Smoking_Status'].map({'Yes': 1, 'No': 0}).fillna(0)  # Default to 0 if unknown
df1['Genetic_Mutation'] = df1['Genetic_Mutation'].map({'Yes': 1, 'No': 0}).fillna(0)
df1['Family_History'] = df1['Family_History'].map({'Yes': 1, 'No': 0}).fillna(0)
df1['Radiation_Exposure'] = df1['Radiation_Exposure'].map({'Yes': 1, 'No': 0}).fillna(0)
df1['Infection_History'] = df1['Infection_History'].map({'Yes': 1, 'No': 0}).fillna(0)
df1['Leukemia_Status'] = df1['Leukemia_Status'].map({'Positive': 1, 'Negative': 0}).fillna(0)
```

```python
df1.head()
```

```python
selected_columns = [
    'Age',
    'Gender',
    'WBC_Count',
    'RBC_Count',
    'Platelet_Count',
    'Hemoglobin_Level',
    'Bone_Marrow_Blasts',
    'Family_History',
    'Smoking_Status',
    'Radiation_Exposure',
    # 'Chronic_Illness',
    # 'Immune_Disorders',
    'BMI',
    'Infection_History'
]
df = df1[selected_columns]
out = df1['Leukemia_Status']
df.head()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import pandas as pd
import numpy as np
```

```python
# Assuming you have your features in X and target in y
# X, y = load_your_data() 

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(df, out, test_size=0.2)
```

```python
# For Classification
rf_model_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model_clf.fit(X_train, y_train)
```

```python
# Make predictions on the test set
y_pred = rf_model_clf.predict(X_test)

# Evaluate the model (e.g., for classification)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")
```

```python
from joblib import dump
import os

# Create the 'model' folder if it doesn't exist
os.makedirs('model', exist_ok=True)

# Save the model to the 'model' folder
dump(rf_model_clf, 'model/random_forest_model.joblib')

print("Model saved to 'model/random_forest_model.joblib'")
```
